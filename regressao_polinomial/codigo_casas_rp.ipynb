```python
import pandas as pd
import numpy as np
```

```python
base_casas = pd.read_csv("D:/cursos/curso-machine-learning/bases_dados/house_prices.csv")
base_casas
```

```python
x_casas = base_casas.iloc[:, 3:19].values
x_casas
```

```python
y_casas = base_casas.iloc[:,2].values
y_casas
```

```python
from sklearn.model_selection import train_test_split
x_casas_treinamento, x_casas_teste, y_casas_treinamento, y_casas_teste = train_test_split(x_casas, y_casas, test_size=0.3, random_state=5)
```

```python
x_casas_treinamento.shape, x_casas_teste.shape
```

```python
y_casas_treinamento.shape
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree = 2)
x_casas_treinamento_poly = poly.fit_transform(x_casas_treinamento)
x_casas_teste_poly = poly.fit_transform(x_casas_teste)
```

```python
x_casas_treinamento_poly.shape, x_casas_teste_poly.shape
```

```python
from sklearn.linear_model import LinearRegression
regressor_casas_poly = LinearRegression()
regressor_casas_poly.fit(x_casas_treinamento_poly, y_casas_treinamento)
```

```python
regressor_casas_poly.score(x_casas_treinamento_poly, y_casas_treinamento)
```

```python
regressor_casas_poly.score(x_casas_teste_poly, y_casas_teste)
```

```python
previsoes = regressor_casas_poly.predict(x_casas_teste_poly)
previsoes
```

```python
y_casas_teste
```

```python
abs(y_casas_teste - previsoes).mean()
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mean_squared_error(y_casas_teste, previsoes)
```
