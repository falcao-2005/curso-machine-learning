```python
import pandas as pd
import numpy as np
from apyori import apriori
```

```python
base_mercado1 = pd.read_csv('D:/cursos/curso-machine-learning/bases_dados/mercado.csv', header=None)
base_mercado1
```

```python
transacoes = []
for i in range(len(base_mercado1)):
    transacoes.append([str(base_mercado1.values[i, j]) for j in range(base_mercado1.shape[1])])
```

```python
transacoes, type(transacoes)
```

```python
regras = apriori(transacoes, min_support = 0.3, min_confidence = 0.8, min_lift = 2)
resultados = list(regras)
resultados
```

```python
lista_condicao = []
lista_resultado = []
suporte = []
confianca = []
lift = []

for resultado in resultados:
    s = resultado[1]
    #print(s)
    result_rules = resultado[2]
    #print(result_rule)
    for result_rule in result_rules:
        #print(result_rule)  
        a = list(result_rule[0])
        b = list(result_rule[1])
        c = result_rule[2]
        l = result_rule[3]
        #print(a, ' - ', b, ' - ', c, ' - ', l)
        lista_condicao.append(a)
        lista_resultado.append(b)
        confianca.append(c)
        lift.append(l)
        suporte.append(s)
```
