# Detectando a face das pessoas

```python
import matplotlib.pyplot as plt
import cv2

# Carrega a imagem
imagem = cv2.imread('D:/cursos/curso-machine-learning/imagens/workplace-1245776_1920.jpg')

if imagem is not None:
    # O OpenCV lê em BGR, mas o Matplotlib exibe em RGB. Precisamos converter:
    imagem_rgb = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)
    
    plt.imshow(imagem_rgb)
    plt.axis('off') # Esconde as coordenadas do gráfico
    plt.show()
else:
    print("Imagem não encontrada!")
```

```python
detector_face = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)
    
plt.imshow(imagem_cinza, cmap='gray')
plt.axis('off') # Esconde as coordenadas do gráfico
plt.show()
```

```python
deteccoes = detector_face.detectMultiScale(imagem_cinza, scaleFactor=1.35, minSize=(30,30))
```

```python
deteccoes
```

```python
len(deteccoes)
```

```python
for (x, y, l, a) in deteccoes:
    #print(x, y, l, a)
    cv2.rectangle(imagem_rgb, (x, y), (x + l, y + a), (0,255,0), 2)
plt.imshow(imagem_rgb)
plt.axis('off') # Esconde as coordenadas do gráfico
plt.show()
```

# Detectando o corpo inteiro das pessoas

```python
imagem_corpo = cv2.imread('D:/cursos/curso-machine-learning/imagens/pessoas.jpg')
```

```python
plt.imshow(imagem_corpo)
plt.axis('off')
plt.show()
```

```python
detector_corpo = cv2.CascadeClassifier('fullbody.xml')
```

```python
imagem_corpo_cinza = cv2.cvtColor(imagem_corpo, cv2.COLOR_BGR2GRAY)

plt.imshow(imagem_corpo_cinza, cmap='gray')
plt.axis('off')
plt.show()
```

```python
deteccoes = detector_corpo.detectMultiScale(imagem_corpo_cinza, minSize=(20,20))
```

```python
deteccoes
```

```python
len(deteccoes)
```

```python
for (x,y,l,a) in deteccoes:
    cv2.rectangle(imagem_corpo, (x, y), (x + l, y + a), (0,255,0), 2)
plt.imshow(imagem_corpo)
plt.axis('off')
plt.show()
```
