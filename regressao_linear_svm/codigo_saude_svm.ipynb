```python
import pandas as pd
import numpy as np
```

```python
base_plano = pd.read_csv('D:/cursos/curso-machine-learning/bases_dados/plano_saude2.csv')
base_plano
```

```python
x_plano = base_plano.iloc[:,0].values
x_plano = x_plano.reshape(-1,1)
x_plano
```

```python
y_plano = base_plano.iloc[:,1].values
y_plano
```

```python
# Kernel Linear

from sklearn.svm import SVR
regressor_svr = SVR(kernel='linear')
regressor_svr.fit(x_plano, y_plano)
```

```python
import plotly.express as px
```

```python
grafico = px.scatter(x = x_plano.ravel(), y = y_plano)
grafico.add_scatter(x = x_plano.ravel(), y = regressor_svr.predict(x_plano), name = 'Regressao SVR')
grafico.show()
```

# Tem um comportamento similar à regressão linear simples

```python
# Kerenel Polinomial

regressor_svr_poly = SVR(kernel='poly', degree=3)
regressor_svr_poly.fit(x_plano, y_plano)
```

```python
grafico = px.scatter(x = x_plano.ravel(), y = y_plano)
grafico.add_scatter(x = x_plano.ravel(), y = regressor_svr_poly.predict(x_plano), name = 'Regressao SVR')
grafico.show()
```

```python
# Kernel RBF

regressor_svr_rbf = SVR(kernel='rbf')
regressor_svr_rbf.fit(x_plano, y_plano)
```

```python
grafico = px.scatter(x = x_plano.ravel(), y = y_plano)
grafico.add_scatter(x = x_plano.ravel(), y = regressor_svr_rbf.predict(x_plano), name = 'Regressao SVR')
grafico.show()
```

# Resultado estranho porque os dados não estão normalizados

```python
from sklearn.preprocessing import StandardScaler
scaler_x = StandardScaler()
x_plano_scaled = scaler_x.fit_transform(x_plano)

scaler_y = StandardScaler()
y_plano_scaled = scaler_y.fit_transform(y_plano.reshape(-1,1))
```

```python
x_plano_scaled
```

```python
y_plano_scaled
```

```python
regressor_svr_rbf.fit(x_plano_scaled, y_plano_scaled.ravel())
```

```python
grafico = px.scatter(x = x_plano_scaled.ravel(), y = y_plano_scaled.ravel())
grafico.add_scatter(x = x_plano_scaled.ravel(), y = regressor_svr_rbf.predict(x_plano_scaled), name = 'Regressao SVR')
grafico.show()
```
