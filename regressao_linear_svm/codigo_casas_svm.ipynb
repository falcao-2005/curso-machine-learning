```python
import pandas as pd
import numpy as np
```

```python
base_casas = pd.read_csv('D:/cursos/curso-machine-learning/bases_dados/house_prices.csv')
base_casas
```

```python
x_casas = base_casas.iloc[:,3:21].values
x_casas
```

```python
x_casas.shape
```

```python
y_casas = base_casas.iloc[:,2].values
y_casas
```

```python
y_casas.shape
```

```python
from sklearn.model_selection import train_test_split
x_casas_train, x_casas_test, y_casas_train, y_casas_test = train_test_split(x_casas, y_casas, test_size=0.3, random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
scaler_x = StandardScaler()
x_casas_train_scaled = scaler_x.fit_transform(x_casas_train)

scaler_y = StandardScaler()
y_casas_train_scaled = scaler_y.fit_transform(y_casas_train.reshape(-1,1))
```

```python
x_casas_test_scaled = scaler_x.fit_transform(x_casas_test)
y_casas_test_scaled = scaler_y.fit_transform(y_casas_test.reshape(-1,1))
```

```python
# kernel rbf

from sklearn.svm import SVR
regressor_svr = SVR(kernel='rbf')
regressor_svr.fit(x_casas_train_scaled, y_casas_train_scaled.ravel())
```

```python
previsoes = regressor_svr.predict(x_casas_test_scaled)
previsoes
```

```python
regressor_svr.score(x_casas_train_scaled, y_casas_train_scaled)
```

```python
regressor_svr.score(x_casas_test_scaled, y_casas_test_scaled)
```

```python
inverse_previsao = scaler_y.inverse_transform(previsoes.reshape(-1,1))
```

```python
inverse_y_scaled = scaler_y.inverse_transform(y_casas_test_scaled)
```

```python
from sklearn.metrics import mean_absolute_error
mean_absolute_error(inverse_y_scaled, inverse_previsao)
```
